```python
import torch
import torch.nn as nn

# BCE 多标签分类
# 使用BCEWithLogitsLoss处理多标签分类问题
bce_loss = nn.BCEWithLogitsLoss()
logits = torch.tensor([0.8, -1.2, 0.3])  # 对应滑坡、崩塌、泥石流
labels = torch.tensor([1.0, 0.0, 1.0])    # 表示滑坡+泥石流
loss = bce_loss(logits, labels)
print("BCE Loss:", loss.item())

# CrossEntropy 单标签分类
# 使用CrossEntropyLoss处理单标签分类问题
ce_loss = nn.CrossEntropyLoss()
logits = torch.tensor([[2.1, -0.5, 0.3]])  # 三类灾害打分
label = torch.tensor([0])  # 表示是类别0（滑坡）
loss = ce_loss(logits, label)
print("CrossEntropy Loss:", loss.item())
```
